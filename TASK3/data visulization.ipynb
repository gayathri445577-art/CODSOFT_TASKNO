```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data={"product":["laptop","phone","headphones","smartwatch","tablet","ipad"],"sales":[120,200,300,45,90,110]}
df=pd.DataFrame(data)
print(df)
plt.bar(df["product"],df["sales"])
plt.title("salary by product")
plt.xlabel("product")
plt.ylabel("sales")
plt.show()
```

```python
 plt.plot(df["product"],df["sales"],marker="o")
 plt.title("sales by product")
 plt.xlabel("product")
 plt.ylabel("sales")
 plt.show()
```

```python
plt.pie(df["sales"],labels=df['product'],autopct="%1.1f%%")
plt.title("sales distibution by product")
plt.show()
```

```python
plt.hist(df['sales'])
plt.title("sales distribution")
plt.xlabel('sales')
plt.ylabel("frequency")
plt.show()
```

```python
plt.scatter(df["product"],df["sales"])
plt.title("sales scatter plot")
plt.xlabel("product")
plt.ylabel("sales")
plt.show()
```

```python
print("highest sales:",df.loc[df["sales"].idxmax(),"product"])
print("lowest sales:",df.loc[df["sales"].idxmin(),"product"])
print("average sales:",df["sales"].mean())
```

```python
fig,axes=plt.subplots(2,2,figsize=(12,8))
#bar chart
axes[0,0].bar(df["product"],df["sales"])
axes[0,0].set_title("sales by product")
axes[0,0].tick_params(axis="x",rotation=45)
#line chart
axes[0,1].plot(df["product"],df["sales"],marker="o")
axes[0,1].set_title("sales trend")
axes[0,1].tick_params(axis="x",rotation=45)
#pie chart
axes[1,0].pie(df["sales"],labels=df["product"],autopct="%1.1f%%")
axes[1,0].set_title("sales distribution")
#histogram
axes[1,1].hist(df["sales"])
axes[1,1].set_title("sales distribution")
plt.tight_layout()
plt.show()
```
